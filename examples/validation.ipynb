```python
import pandas as pd 
import numpy as np 
```

```python
import pyecoacc as acc 
from pyecoacc.models.pipeline import get_default_random_forest_pipeline
from pyecoacc.util.analytics import compare_models_cv
```

# Load 

```python
molerats_data = pd.read_csv("data/molerats.csv", index_col=0)
```

```python
animal_identiy = molerats_data["Animal"].values 
molerats_data.drop("Animal", inplace=True, axis=1)

X = molerats_data.iloc[:, :-1].values 
y = molerats_data.Behavior.values 
```

# Validation #1) Stratified cross-validation
Tries to split the observations of each behavior equally between the folds. 

```python
model = get_default_random_forest_pipeline()
model_dict = {"random forest": model}
```

```python
accuracy, *_ = compare_models_cv(X, y, model_dict, cv=5, cv_method="stratified")
```

```python
accuracy
```

# Validation #2) Group k-fold cross-validation. 

All samples of an individual are either in the training data or the test data in each fold. This is the correct estimation procedure for model performance when it is going to be used on different animals from the training data.

```python
accuracy, *_ = compare_models_cv(X, y, model_dict, cv_method="animal-groups", individuals=animal_identiy)
```

```python
accuracy
```

# Validation #3) Leave one animal out cross validation. 
This is the extreme version of (2) where k is equal the number of animals. 

```python
accuracy, *_ = compare_models_cv(X, y, model_dict, cv_method="LOIO", individuals=animal_identiy)
```

```python
accuracy
```
